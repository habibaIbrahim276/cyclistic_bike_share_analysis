----------------------------------------------- Staging --------------------------------------------

This notebook contains the data preparation and staging process for Cyclistic bike analysis project 

Steps: 
- Loading and merging multiple CSV files into one dataset 
- Exploring data structure and checking data types 
- Converting data columns to datetime format
- Creating new features such as ride_length, day_of_week, month, and hour
- Checking and handlling missing values 
- Verifying and removing duplicate records
- Saving a cleaned dataseet for further analysis 
This stage ensures that the data is consistent, clean, and suitable for exploratory data analysis in next phase

Merging data

```python
df = pd.concat([may_2025, jun_2025, jul_2025, aug_2025, sep_2025, oct_2025, nov_2025, dec_2025, jan_2026, feb_2026, mar_2026, apr_2026], ignore_index=True)
```

```python
df.head()
```

-------------------------------------------------------------------------------------------------------------------------------------------------

Explore date 

```python
df.columns
```

```python
df.info()
```

Changing data types

```python
df['started_at'] = pd.to_datetime(df['started_at'])
df['ended_at'] = pd.to_datetime(df['ended_at'])
```

Creating Important columns

```python
# ride length 
df['ride_length'] = pd.to_datetime(df['ended_at']) - pd.to_datetime(df['started_at'])
```

```python
# day of week 
df['day_of_week'] = df['started_at'].dt.day_name()
```

```python
df['month'] = df['started_at'].dt.month_name()
```

```python
df['hour'] = df['started_at'].dt.hour
```

```python
df.head()
```

```python
df['month'].unique()
```

Check duplicates 

```python
df.duplicated().sum()
```

Check null values

```python
df.isna().sum()
```

```python
df = df.dropna(subset=['end_lat', 'end_lng'])
```

```python
df.isna().sum()
```

Creating a copy of data that don't have nulls 

```python
df_without_nulls = df.dropna()
```
